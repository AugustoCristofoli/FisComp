```python
"""
Definição da função a ser calculada, necessario mudar apenas isso. A derivada é calculada numericamente
"""
def f(x):
    return x**3 - 2*x +2 

def derivative(f,x, dx=1e-10):
    return (f(x+dx) - f(x-dx))/(2*dx)

def n_r(f, df, x0, max_iter, eps=1e-4):
    x = x0
    for i in range(max_iter):
        if df(f, x) == 0:
            return x, False
        x_new = x - (f(x) / df(f, x))
        if abs(x_new - x) < eps:
            return x, True
        x = x_new
    else:
        return x, False
    

vals = [1+0j, -1+1j, -1-1j, 0.5+0.5j, 0+0j]
for x0 in vals:
    (root, converge) = n_r(f, derivative, x0, int(1e5))
    print(f"Raiz: {root} | Converge? {converge}")
```
