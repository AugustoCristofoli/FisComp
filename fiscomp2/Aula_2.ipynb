N-R em um intervalo

```python
def f(x):
    return (x-4) * (x+3) 

def df(x):
    return 2*x -1

def derivative(f,x, dx=1e-10):
    return (f(x+dx) - f(x-dx))/(2*dx)

def n_r(f, df, x0, eps=1e-4):
    x = x0
    for i in range(1000):
        x_new = x - (f(x) / df(f, x))
        if abs(x_new - x) < eps:
            return x, f(x)
        x = x_new
    return 0

for i in range(-5,5):
    (root, f_val) = n_r(f, derivative, i)
    print(f"Valor Inicial: {i} | Raiz: {round(root)} | Valor de f: {f_val}")
```
